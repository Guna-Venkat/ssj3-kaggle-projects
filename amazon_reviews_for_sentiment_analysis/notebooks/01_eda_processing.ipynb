```python
import pandas as pd
import re
import string

# 🔹 Clean text function
def clean_text(text):
    text = text.lower()
    text = re.sub(r"http\S+", "", text)  # Remove URLs
    text = re.sub(r"\d+", "", text)      # Remove numbers
    text = text.translate(str.maketrans('', '', string.punctuation))  # Remove punctuation
    text = re.sub(r"\s+", " ", text).strip()  # Remove extra spaces
    return text

# 🔹 Load and preprocess function
def preprocess_txt_file(input_path, output_path):
    data = []

    with open(input_path, 'r', encoding='utf-8') as file:
        for line in file:
            if line.strip():
                label, text = line.strip().split(' ', 1)
                label = 0 if label == '__label__1' else 1  # 0 = negative, 1 = positive
                cleaned_text = clean_text(text)
                data.append((label, cleaned_text))

    # Create DataFrame and save
    df = pd.DataFrame(data, columns=["label", "text"])
    df.to_csv(output_path, index=False)
    print(f"✅ Preprocessed data saved to: {output_path}")
    print(df.head())

# 🔧 Run this
preprocess_txt_file("C:\\Users\\gunav\\Downloads\\ssj3-kaggle-projects\\amazon_reviews_for_sentiment_analysis\\data\\raw\\train.ft.txt", "C:\\Users\\gunav\\Downloads\\ssj3-kaggle-projects\\amazon_reviews_for_sentiment_analysis\\data\\preprocessed\\train.csv")
preprocess_txt_file("C:\\Users\\gunav\\Downloads\\ssj3-kaggle-projects\\amazon_reviews_for_sentiment_analysis\\data\\raw\\test.ft.txt", "C:\\Users\\gunav\\Downloads\\ssj3-kaggle-projects\\amazon_reviews_for_sentiment_analysis\\data\\preprocessed\\test.csv")
```
